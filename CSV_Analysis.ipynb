```python
import pandas as pd

# Load the CSV file
df = pd.read_csv('StudentsPerformance.csv')

# First 5 rows
print("First 5 rows:")
print(df.head())

# Shape (rows, columns)
print("\nDataset Shape:", df.shape)

# Column names
print("\nColumns:", df.columns.tolist())

# Info (data types, nulls)
print("\nDataset Info:")
df.info()

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())
```

```python
avg_math = df['math score'].mean()
avg_reading = df['reading score'].mean()
avg_writing = df['writing score'].mean()

print(f"Average Math Score: {avg_math:.2f}")
print(f"Average Reading Score: {avg_reading:.2f}")
print(f"Average Writing Score: {avg_writing:.2f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
gender_avg = df.groupby('gender')['math score'].mean()

plt.figure(figsize=(6,4))
plt.bar(gender_avg.index, gender_avg.values, color=['#FF6B6B', '#4ECDC4'])
plt.title('Average Math Score by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Math Score')
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df['math score'], df['reading score'], alpha=0.5, color='purple')
plt.title('Math Score vs Reading Score')
plt.xlabel('Math Score')
plt.ylabel('Reading Score')
plt.show()
```

```python
# Calculate correlation matrix
corr_matrix = df[['math score', 'reading score', 'writing score']].corr()

plt.figure(figsize=(6,5))
plt.imshow(corr_matrix, cmap='coolwarm', interpolation='none')
plt.colorbar()
plt.xticks(range(len(corr_matrix.columns)), corr_matrix.columns, rotation=45)
plt.yticks(range(len(corr_matrix.columns)), corr_matrix.columns)

# Add correlation values as text on the heatmap
for i in range(len(corr_matrix.columns)):
    for j in range(len(corr_matrix.columns)):
        plt.text(j, i, f'{corr_matrix.iloc[i, j]:.2f}', ha='center', va='center', color='black')

plt.title('Correlation Heatmap of Scores')
plt.tight_layout()
plt.show()
```

Insights:
1. Math and Reading scores show a strong positive correlation of 0.82, indicating that students who perform well in math tend to perform well in reading as well.
2. Math and Writing scores have a correlation of 0.80, showing a similarly strong relationship.
3. Reading and Writing scores have the highest correlation at 0.95, suggesting that language-based skills (reading and writing) are very closely related to each other.
4. Overall, the data shows that students who excel in one subject generally tend to excel in others as well, with reading and writing skills being the most tightly linked.
